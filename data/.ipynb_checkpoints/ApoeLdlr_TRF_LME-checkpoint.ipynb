```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
import pandas as pd
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.tools.sm_exceptions import ConvergenceWarning
```

# Longitudinal Food Consumption

```python
long_consumpt = pd.read_csv('./mouseweights_long.csv', index_col=0)
long_consumpt.head()
```

```python
long_consumpt2 = long_consumpt.dropna(subset=['weight_g'])
```

```python
apoe_long_consumpt = long_consumpt2[long_consumpt2['genotype'] == 'ApoeKO']
ldlr_long_consumpt = long_consumpt2[long_consumpt2['genotype'] == 'LdlrKO']
```

```python
apoe_long_LME = smf.mixedlm("weight_g ~ feeding*week", 
                    apoe_long_consumpt, 
                    groups = apoe_long_consumpt["group"]) 
#does not include re_formula = random effects
```

```python
apoe_long_LME_fit = apoe_long_LME.fit()
print(apoe_long_LME_fit.summary())
```

```python
sns.lmplot(x="week", y="weight_g", hue='feeding', data=apoe_long_consumpt)
```

```python
#Residuals, should be roughly zero-centered and normal
def pltcolor(lst):
    cols=[]
    for l in lst:
        if l=='adlib':
            cols.append('brown')
        elif l=='TRF':
            cols.append('green')
        else:
            cols.append('black')
    return cols

cols=pltcolor(apoe_long_consumpt.feeding.values)

plt.scatter(apoe_long_consumpt['weight_g'] - apoe_long_LME_fit.resid, apoe_long_LME_fit.resid, c=cols, alpha = 0.5)
plt.axhline(0)
plt.title("Residual vs. Fitted in Python")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
#plt.savefig('python_plot.png',dpi=300)
plt.show()
```

```python
ldlr_long_LME = smf.mixedlm("weight_g ~ feeding*week", 
                    ldlr_long_consumpt, 
                    groups = ldlr_long_consumpt["group"]) 
#does not include re_formula = random effects
```

```python
ldlr_long_LME_fit = ldlr_long_LME.fit()
print(ldlr_long_LME_fit.summary())
```

```python
sns.lmplot(x="week", y="weight_g", hue='feeding', data=ldlr_long_consumpt)
```

```python
#Residuals, should be roughly zero-centered and normal
def pltcolor(lst):
    cols=[]
    for l in lst:
        if l=='adlib':
            cols.append('brown')
        elif l=='TRF':
            cols.append('green')
        else:
            cols.append('black')
    return cols

cols=pltcolor(ldlr_long_consumpt.feeding.values)

plt.scatter(ldlr_long_consumpt['weight_g'] - ldlr_long_LME_fit.resid, ldlr_long_LME_fit.resid, c=cols, alpha = 0.5)
plt.axhline(0)
plt.title("Residual vs. Fitted in Python")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
#plt.savefig('python_plot.png',dpi=300)
plt.show()
```

## Metabolic Cage Food Consumption

```python
consumption = pd.read_csv('./MetabolicCage/foodconsumption_formatted.csv', index_col=0)
consumption.head()
```

```python
consumption.food_change_g.describe()
```

```python
consumption2 = consumption.dropna() #remove nans
consumption2 = consumption2[consumption2['food_change_g'] > 0] #remove 0s
consumption2['log_foodchange'] = np.log(consumption2['food_change_g']) #log transformation
```

```python
consumption2.tail()
```

```python
apoe_consumption = consumption2[consumption2['genotype'] == 'ApoeKO']
ldlr_consumption = consumption2[consumption2['genotype'] == 'LdlrKO']
```

```python
apoe_mixed = smf.mixedlm("log_foodchange ~ feeding*ZT_bin", 
                    apoe_consumption, 
                    groups = apoe_consumption["group"]) 
#does not include re_formula = random effects
```

```python
apoe_mixed_fit = apoe_mixed.fit()
print(apoe_mixed_fit.summary())
```

```python
sns.lmplot(x="ZT_bin", y="log_foodchange", hue='feeding', data=apoe_consumption)
```

```python
#Residuals, should be roughly zero-centered and normal
def pltcolor(lst):
    cols=[]
    for l in lst:
        if l=='adlib':
            cols.append('brown')
        elif l=='TRF':
            cols.append('green')
        else:
            cols.append('black')
    return cols

cols=pltcolor(apoe_consumption.feeding.values)

plt.scatter(apoe_consumption['log_foodchange'] - apoe_mixed_fit.resid, apoe_mixed_fit.resid, c=cols, alpha = 0.5)
plt.axhline(0)
plt.title("Residual vs. Fitted in Python")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
#plt.savefig('python_plot.png',dpi=300)
plt.show()
```

```python
ldlr_mixed = smf.mixedlm("log_foodchange ~ feeding*ZT_bin", 
                    ldlr_consumption, 
                    groups = ldlr_consumption["group"]) 
#does not include re_formula = random effects
```

```python
ldlr_mixed_fit = ldlr_mixed.fit()
print(ldlr_mixed_fit.summary())
```

```python
sns.lmplot(x="ZT_bin", y="log_foodchange", hue='feeding', data=ldlr_consumption)
```

```python
#Residuals, should be roughly zero-centered and normal
def pltcolor(lst):
    cols=[]
    for l in lst:
        if l=='adlib':
            cols.append('brown')
        elif l=='TRF':
            cols.append('green')
        else:
            cols.append('black')
    return cols

cols=pltcolor(ldlr_consumption.feeding.values)

plt.scatter(ldlr_consumption['log_foodchange'] - ldlr_mixed_fit.resid, ldlr_mixed_fit.resid, c=cols, alpha = 0.5)
plt.axhline(0)
plt.title("Residual vs. Fitted in Python")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
#plt.savefig('python_plot.png',dpi=300)
plt.show()
```
