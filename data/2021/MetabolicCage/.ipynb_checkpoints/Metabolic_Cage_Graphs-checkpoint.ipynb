# Redo Food Consumption

```python
food = pd.read_csv('./fooduptake_raw.csv')
food.head()
```

```python
food['Date_Time_1'] = pd.to_datetime(food['Date_Time_1'])
```

```python
food.head()
```

```python
food.columns
```

```python
for cols in ['123N', '124L', '124N', '123L', '125L', 'died', '126N', '126L', '130N', '130R', 
             '131N', '131L', '132L', '133N', '133L', '133R']:
    foodchange = food[cols].diff()
```

```python
foodchange = food.copy()
foodchange.set_index(['Date_Time_1', 'ZT Time'], inplace=True)
foodchange.head()
```

```python
foodchanged = foodchange.diff()
foodchanged.head()
```

```python
refoodchanged = foodchanged.T
refoodchanged
```

```python
unfoodchanged = foodchanged.reset_index()
unfoodchanged
```

```python
Tfoodchanged = unfoodchanged.melt(value_vars=['123N', '124L', '124N', '123L', '125L', 'died', '126N', '126L', '130N', '130R', 
             '131N', '131L', '132L', '133N', '133L', '133R'], id_vars=['Date_Time_1', 'ZT Time'], 
                                 var_name='mouse', value_name='food_change_g')
Tfoodchanged
```

```python
key = pd.read_csv('./mouse_key.csv')
key.head()
```

```python
consumption = pd.merge(Tfoodchanged, key, left_on='mouse', right_on='mouse')
consumption
```

```python
consumption['hour_bin'] = [r.hour for r in consumption.Date_Time_1]
```

```python
consumption.head()
```

```python
consumption.hour_bin.unique()
```

```python
def map_values(row, values_dict):
    return values_dict[row]

values_dict = { 18: 0, 19: 1, 20: 2, 21: 3, 22: 4, 23: 5, 0: 6, 1: 7, 2: 8, 3: 9, 4: 10, 5: 11, 6: 12, 
              7: 13, 8: 14, 9: 15, 10: 16, 11: 17, 12: 18, 13: 19, 14: 20, 15: 21, 16: 22, 17: 23}

consumption['ZT_bin'] = consumption['hour_bin'].apply(map_values, args = (values_dict,))
```

```python
consumption.head()
```

```python
consumption.ZT_bin.unique()
```

```python
#1 hr blocks
sns.set_context("paper")
sns.set_style("white")
g1 = sns.relplot(x="ZT_bin", y="food_change_g", hue="group", col="genotype", col_order=["ApoeKO", "LdlrKO"],
             palette=["forestgreen", "saddlebrown", "forestgreen", "saddlebrown"], kind="line", ci=68, data=consumption)
#g1.set(xticks=consumption.ZT_bin.values)

#axes = g1.axes.flatten()
#for ax in axes:
#    ax.set_xlabel("Zeitgeber Time (ZT)")
    
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = 'Arial'
plt.rcParams['svg.fonttype'] = 'none'
plt.savefig('../../Figures/food_consumption_avg2.svg', dpi=300, bbox_inches="tight")
```

```python
#5 min blocks
sns.set_context("paper")
sns.set_style("white")
g1 = sns.relplot(x="ZT Time", y="food_change_g", hue="group", col="genotype", col_order=["ApoeKO", "LdlrKO"],
             palette=["forestgreen", "saddlebrown", "forestgreen", "saddlebrown"], kind="line", ci=68, data=consumption)
    
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = 'Arial'
plt.rcParams['svg.fonttype'] = 'none'
#plt.savefig('../../Figures/food_consumption_avg2.svg', dpi=300, bbox_inches="tight")
```

```python
consumption.to_csv('./foodconsumption_formatted.csv')
```
